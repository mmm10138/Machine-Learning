# Part 1: Solving ODEs with PINNs

## Training functions

```python
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Define the Neural Network model
class NeuralNet(nn.Module):
    def __init__(self):
        super(NeuralNet, self).__init__()
        self.fc1 = nn.Linear(1, 200)  # First layer (input -> 200 neurons)
        self.fc2 = nn.Linear(200, 200)  # Second layer (200 -> 200 neurons)
        self.fc3 = nn.Linear(200, 1)  # Output layer (200 -> 1 neuron)

    def forward(self, t):
        x = torch.sigmoid(self.fc1(t))  # Sigmoid activation for the hidden layers
        x = torch.sigmoid(self.fc2(x))  # Sigmoid activation for the hidden layers
        x = self.fc3(x)  # No activation on the output layer
        return x
```

```python
# Define the true solution of the ODE
def true_solution(t):
    return t - 1 + 2 * torch.exp(-t)

# Define the loss function for the ODE residual
def ode_loss(model, t_batch):
    t_batch.requires_grad_(True)
    y_pred = model(t_batch)

    # Compute dy/dt using autograd
    dy_dt = torch.autograd.grad(y_pred, t_batch, grad_outputs=torch.ones_like(y_pred), create_graph=True)[0]

    # Residual of the ODE: dy/dt = t - y
    # i.e., dy(t)/dt = f(t, y(t)) where f(t,y) = t - y
    residual = dy_dt - (t_batch - y_pred)

    # MSE loss for the ODE residual
    loss_ode = torch.mean(residual ** 2)

    return loss_ode

# Define the loss for the initial condition y(0) = 1
def initial_condition_loss(model):
    y0_pred = model(torch.tensor([[0.0]]))  # y(0)
    loss_ic = (y0_pred - 1.0) ** 2  # (y(0) - 1)^2
    return loss_ic

# Define the loss using real data
def data_loss(model, t_data, y_data):
  y_pred = model(t_data)
  return torch.mean((y_pred - y_data) ** 2)
```

```python
# Main function to train the model
def train_ode_solver(n_data=5, w_data=1, lr=0.001, num_epochs=1000, batch_size=10, test_every=20):
    # Initialize the model
    model = NeuralNet()

    # Define the optimizer
    optimizer = optim.Adam(model.parameters())

    # Define the time range
    T = 5.0

    # Create true data points
    t_data = T * torch.rand(n_data, 1)
    y_data = true_solution(t_data).detach()

    # Create a testing set of 100 points sampled uniformly from [0, T]
    # to measure the error
    t_error = torch.linspace(0, T, 100).unsqueeze(1)
    y_error_true = true_solution(t_error).detach()

    # Lists to store losses
    train_losses = []
    test_errors = []

    # Training loop
    for epoch in range(num_epochs):
        model.train()

        # Sample minibatch of time points uniformly in [0, T]
        t_batch = T * torch.rand(batch_size, 1)

        # Compute the total loss: ODE residual loss + initial condition loss
        loss_ode = ode_loss(model, t_batch)
        loss_ic = initial_condition_loss(model)
        loss_data = data_loss(model, t_data, y_data)
        loss = loss_ode + loss_ic + w_data*loss_data

        # Backpropagation and optimizer step
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        # Record train loss
        train_losses.append(loss.item())

        # Evaluate the model on the testing set every few epochs
        if epoch % test_every == 0:
            model.eval()
            with torch.no_grad():
                y_error_pred = model(t_error)
                test_error = torch.mean((y_error_pred - y_error_true) ** 2)
                test_errors.append(test_error.item())
                print(f'Epoch {epoch}, Error: {test_error.item()}')

                # Plot the true solution and the learned solution by the NN
                plt.figure(figsize=(5, 5))
                plt.plot(t_error.numpy(), y_error_true.numpy(), 'r--', label="True Solution")
                plt.scatter(t_data.numpy(), y_data.numpy(), color='green', marker='x', label="True Data Points")
                plt.plot(t_error.numpy(), y_error_pred.numpy(), 'b-', label="Learned Solution (NN)")
                plt.legend()
                plt.xlabel("t")
                plt.ylabel("y(t)")
                plt.title(f"Epoch {epoch}")
                plt.grid(True, which="both", ls="--")
                plt.show()

    # Plot train loss and errors
    plt.figure(figsize=(8, 6))
    plt.plot(np.arange(num_epochs), train_losses, label='Train Loss')
    plt.plot(np.arange(0, num_epochs, test_every), test_errors, label='Test MSE (vs. True Solution)', color='red')
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.title("Train Loss and Errors Over Epochs")
    plt.legend()
    plt.grid(True, which="both", ls="--")
    plt.show()

    return model, train_losses, test_errors
```

## First test

```python
# Run the training
np.random.seed(42)
torch.manual_seed(42)
_, _, _ = train_ode_solver()
```

## Larger weight on data loss

If we give too much weight to the data points, we increase the risk of overfitting: the neural network will stick to the (few) data points instead of trying to solve the ODE over the whole interval.

```python
# Run the training
np.random.seed(42)
torch.manual_seed(42)
_, _, _ = train_ode_solver(w_data=100)
```

# Part 2: Inverse problem

## Training functions

```python
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
```

```python
# Define the Neural Network model
class NeuralNet(nn.Module):
    def __init__(self):
        super(NeuralNet, self).__init__()
        self.fc1 = nn.Linear(1, 200)  # First layer (input -> 200 neurons)
        self.fc2 = nn.Linear(200, 200)  # Second layer (200 -> 200 neurons)
        self.fc3 = nn.Linear(200, 1)  # Output layer (200 -> 1 neuron)

    def forward(self, t):
        x = torch.sigmoid(self.fc1(t))  # Sigmoid activation for the hidden layers
        x = torch.sigmoid(self.fc2(x))  # Sigmoid activation for the hidden layers
        x = self.fc3(x)  # No activation on the output layer
        return x
```

```python
# Define the true solution of the ODE (for testing purposes)
def true_q(t):
    return t

def true_solution(t):
    return true_q(t) - 1 + 2 * torch.exp(-t)

# Define the loss function for the ODE residual with model_y and model_q
def ode_loss(model_y, model_q, t_batch):
    t_batch.requires_grad_(True)
    y_pred = model_y(t_batch)

    # Compute dy/dt using autograd
    dy_dt = torch.autograd.grad(y_pred, t_batch, grad_outputs=torch.ones_like(y_pred), create_graph=True)[0]

    # Residual of the ODE: dy/dt = q(t) - y(t)
    q_pred = model_q(t_batch)
    residual = dy_dt - (q_pred - y_pred)

    # MSE loss for the ODE residual
    loss_ode = torch.mean(residual ** 2)
    return loss_ode

# Define the loss for the initial condition y(0) = 1
def initial_condition_loss(model_y):
    y0_pred = model_y(torch.tensor([[0.0]]))  # y(0)
    loss_ic = (y0_pred - 1.0) ** 2  # (y(0) - 1)^2
    return loss_ic

# Define the loss using real data
def data_loss(model_y, t_data, y_data):
    y_pred = model_y(t_data)
    return torch.mean((y_pred - y_data) ** 2)
```

```python
# Main function to train the models (model_y for y and model_q for q)
def train_ode_solver_q(n_data=5, w_data=1, learning_rate=0.001, num_epochs=1000, batch_size = 10, test_every=20):
    # Initialize the models for y and q
    model_y = NeuralNet()
    model_q = NeuralNet()

    # Define the optimizer for both networks
    optimizer = optim.Adam(list(model_y.parameters()) + list(model_q.parameters()), lr=learning_rate)

    # Define the time range
    T = 5.0

    # Create true data points
    t_data = T * torch.rand(n_data, 1)
    y_data = true_solution(t_data).detach()

    # Create a testing set of 100 points sampled uniformly from [0, T]
    t_error = torch.linspace(0, T, 100).unsqueeze(1)
    y_error_true = true_solution(t_error).detach()

    # Lists to store losses
    train_losses = []
    test_errors = []

    # Training loop
    for epoch in range(num_epochs):
        model_y.train()
        model_q.train()

        # Sample minibatch of time points uniformly in [0, T]
        t_batch = T * torch.rand(batch_size, 1)

        # Compute the total loss: ODE residual loss + initial condition loss + data loss
        loss_ode = ode_loss(model_y, model_q, t_batch)
        loss_ic = initial_condition_loss(model_y)
        loss_data = data_loss(model_y, t_data, y_data)
        loss = loss_ode + loss_ic + w_data * loss_data

        # Backpropagation and optimizer step
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        # Record train loss
        train_losses.append(loss.item())

        # Evaluate the model on the testing set every few epochs
        if epoch % test_every == 0:
            model_y.eval()
            model_q.eval()
            with torch.no_grad():
                y_error_pred = model_y(t_error)
                q_error_pred = model_q(t_error)
                test_error = torch.mean((y_error_pred - y_error_true) ** 2)
                test_errors.append(test_error.item())
                print(f'Epoch {epoch}, Error: {test_error.item()}')

                # Plot the true solution and the learned solution by the NN
                plt.figure(figsize=(10, 5))
                plt.subplot(1, 2, 1)
                plt.plot(t_error.numpy(), y_error_true.numpy(), 'r--', label="True Solution y")
                plt.scatter(t_data.numpy(), y_data.numpy(), color='green', marker='x', label="True Data Points")
                plt.plot(t_error.numpy(), y_error_pred.numpy(), 'b-', label="Learned Solution (y NN)")
                plt.legend()
                plt.xlabel("t")
                plt.ylabel("y(t)")
                plt.title(f"Epoch {epoch}")
                plt.grid(True, which="both", ls="--")

                # Plot q
                plt.subplot(1, 2, 2)
                plt.plot(t_error.numpy(), true_q(t_error.numpy()), 'r--', label="True Solution q")
                plt.plot(t_error.numpy(), q_error_pred.numpy(), 'b-', label="Learned Solution (q NN)")
                plt.legend()
                plt.xlabel("t")
                plt.ylabel("q(t)")
                plt.title(f"Epoch {epoch}")
                plt.grid(True, which="both", ls="--")
                plt.show()

    # Plot train loss and errors
    plt.figure(figsize=(8, 6))
    plt.semilogy(np.arange(num_epochs), train_losses, label='Train Loss')
    plt.semilogy(np.arange(0, num_epochs, test_every), test_errors, label='Test MSE (vs. True Solution)', color='red')
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.title("Train Loss and Errors Over Epochs")
    plt.legend()
    plt.grid(True, which="both", ls="--")
    plt.show()

    return model_y, model_q, train_losses, test_errors
```

## Uniform grid of data points

We now take a uniform grid of points on $[0,T]$. This is different from sampling points uniformly because we choose exactly the locations of the points. This setting could be relevant in real-world scenarios if we are able to choose at which times the measurements are done.

```python
# Main function to train the models (model_y for y and model_q for q)
def train_ode_solver_q_unif(n_data=5, w_data=1, learning_rate=0.001, num_epochs=1000, batch_size = 10, test_every=20):
    # Initialize the models for y and q
    model_y = NeuralNet()
    model_q = NeuralNet()

    # Define the optimizer for both networks
    optimizer = optim.Adam(list(model_y.parameters()) + list(model_q.parameters()), lr=learning_rate)

    # Define the time range
    T = 5.0

    # Create true data points
    t_data = torch.linspace(0, T, n_data)
    t_data = t_data.unsqueeze(1)
    y_data = true_solution(t_data).detach()

    # Create a testing set of 100 points sampled uniformly from [0, T]
    t_error = torch.linspace(0, T, 100).unsqueeze(1)
    y_error_true = true_solution(t_error).detach()

    # Lists to store losses
    train_losses = []
    test_errors = []

    # Training loop
    for epoch in range(num_epochs):
        model_y.train()
        model_q.train()

        # Sample minibatch of time points uniformly in [0, T]
        t_batch = T * torch.rand(batch_size, 1)

        # Compute the total loss: ODE residual loss + initial condition loss + data loss
        loss_ode = ode_loss(model_y, model_q, t_batch)
        loss_ic = initial_condition_loss(model_y)
        loss_data = data_loss(model_y, t_data, y_data)
        loss = loss_ode + loss_ic + w_data * loss_data

        # Backpropagation and optimizer step
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        # Record train loss
        train_losses.append(loss.item())

        # Evaluate the model on the testing set every few epochs
        if epoch % test_every == 0:
            model_y.eval()
            model_q.eval()
            with torch.no_grad():
                y_error_pred = model_y(t_error)
                q_error_pred = model_q(t_error)
                test_error = torch.mean((y_error_pred - y_error_true) ** 2)
                test_errors.append(test_error.item())
                print(f'Epoch {epoch}, Error: {test_error.item()}')

                # Plot the true solution and the learned solution by the NN
                plt.figure(figsize=(8, 6))
                plt.subplot(1, 2, 1)
                plt.plot(t_error.numpy(), y_error_true.numpy(), 'r--', label="True Solution y")
                plt.scatter(t_data.numpy(), y_data.numpy(), color='green', marker='x', label="True Data Points")
                plt.plot(t_error.numpy(), y_error_pred.numpy(), 'b-', label="Learned Solution (y NN)")
                plt.legend()
                plt.xlabel("t")
                plt.ylabel("y(t)")
                plt.title(f"Epoch {epoch}")
                plt.grid(True, which="both", ls="--")

                # Plot q
                plt.subplot(1, 2, 2)
                plt.plot(t_error.numpy(), true_q(t_error.numpy()), 'r--', label="True Solution q")
                plt.plot(t_error.numpy(), q_error_pred.numpy(), 'b-', label="Learned Solution (q NN)")
                plt.legend()
                plt.xlabel("t")
                plt.ylabel("q(t)")
                plt.title(f"Epoch {epoch}")
                plt.grid(True, which="both", ls="--")
                plt.show()

    # Plot train loss and errors
    plt.figure(figsize=(8, 6))
    plt.semilogy(np.arange(num_epochs), train_losses, label='Train Loss')
    plt.semilogy(np.arange(0, num_epochs, test_every), test_errors, label='Test MSE (vs. True Solution)', color='red')
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.title("Train Loss and Errors Over Epochs")
    plt.legend()
    plt.grid(True, which="both", ls="--")
    plt.show()

    return model_y, model_q, train_losses, test_errors
```

```python
# Run the training
np.random.seed(42)
torch.manual_seed(42)
_, _, _, _ = train_ode_solver_q_unif()
```
