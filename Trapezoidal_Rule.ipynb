## Trapezoidal rule


The Trapezoidal Rule for approximating the integral of a function $ f(x) $ over the interval $[a, b]$ using $n$ subintervals is given by:
$$
  \int_a^b f(x) \, dx \approx \frac{h}{2} \left( f(a) + 2 \sum_{i=1}^{n-1} f(x_i) + f(b) \right)
$$
where $ h = \frac{b - a}{n} $ is the step size and $ x_i = a + i h $ for $ i = 1, 2, \dots, n-1 $.

Alternatively, the Trapezoidal Rule can be written as:
$$
  \int_a^b f(x) \, dx \approx \sum_{i=0}^{n-1} \frac{f(x_i) + f(x_{i+1})}{2} h
$$
where $x_i = a + i h $ and $ x_{i+1} = a + (i+1) h $.

```python
import numpy as np

# Trapezoidal rule function
def trapezoidal_rule(a, b, f, n):
    # Step size
    h = (b - a) / n

    # Compute the integral using trapezoidal rule
    integral = 0.5 * (f(a) + f(b))  # First and last terms
    for i in range(1, n):
        x = a+i*h
        integral += f(x)

    # Multiply by the step size
    integral *= h
    return integral

# Test the function with f(x) = sin(x), a=0, b=1, and n=4
a = 0
b = np.pi
n = 4
result = trapezoidal_rule(a, b, np.sin, n)
print(f"Approximate integral of sin(x) from {a} to {b} using {n} intervals: {result}")
```
