## Newton's method

```python
# Newton's method to compute square root of 2
def newtons_method_sqrt2(x0, tol=1e-8, max_iter=100):
    # Function f(x) = x^2 - 2
    f = lambda x: x**2 - 2
    # Derivative f'(x) = 2x
    f_prime = lambda x: 2*x

    x = x0
    for i in range(max_iter):
        # Newton's update formula: x_{n+1} = x_n - f(x_n) / f'(x_n)
        x_new=x-f(x)/f_prime(x)

        # Check if the difference is within tolerance
        if abs(x_new - x) < tol:
            print(f"Converged after {i+1} iterations.")
            return x_new

        # Update x
        x=x_new

    print("Did not converge within the maximum number of iterations.")
    return x

# Starting point
x0 = 1.5
result = newtons_method_sqrt2(x0)
print(f"Approximation of square root of 2: {result}")
print(f"Python's value: {2**0.5}")
```
